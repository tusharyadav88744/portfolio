```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Average Total Bill:", df["total_bill"].mean())
print("Average Tip:", df["tip"].mean())
print("Maximum Bill:", df["total_bill"].max())
print("Minimum Bill:", df["total_bill"].min())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(df["total_bill"], df["tip"])

plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.title("Relationship Between Total Bill and Tip")

plt.grid(True)
plt.show()
```

```python
day_analysis = df.groupby("day")["total_bill"].mean()

print(day_analysis)

day_analysis.plot(kind="bar", figsize=(8, 5))

plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Bill ($)")
plt.xticks(rotation=0)
plt.show()
```

```python
best_day = day_analysis.idxmax()
best_value = day_analysis.max()

print("Highest Average Bill Day:", best_day)
print("Average Bill:", round(best_value, 2))
```
